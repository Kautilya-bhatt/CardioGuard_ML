```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
accuracy_score, precision_score, recall_score,
f1_score,confusion_matrix, classification_report
)
print("Libraries import successfully")
```

```python
#Loading the Dataset
df = pd.read_csv("/content/train.csv")

print("Dataset loaded successfully")
print(df.head())
```

```python
#Understanding the Dataset

#Inspect the dataset
print("First five rows:")
print(df.head())

print("Last five rows:")
print(df.tail())

print("\nDataset shape:")
print(df.shape)

print("\nDataset names:")
print(df.columns)

print("\nDataset info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
# Check Data Quality

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print(df['Heart Disease'].describe())
```

```python
df = df.drop_duplicates()

print("Duplicate rows:", df.duplicated().sum())
print("Final shape:", df.shape)
```

```python
#Graph 1 — Age Vs Heart Disease

plt.figure(figsize=(8, 6))

plt.scatter(df['Age'], df['Heart Disease'])

plt.xlabel('Age')
plt.ylabel('Heart Disease')
plt.title('Representation of Heart Disease in terms of Age')
plt.grid(True)

plt.show()
```

```python
# #Graph 2 — AGE FREQUENCY REPRESENTATION

plt.figure(figsize=(8, 5))

plt.hist(df['Age'], bins=20, edgecolor='black')

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#HEART DISEASE REPESENTATION
df["Heart Disease"].value_counts().sort_index().plot(
    kind="bar",rot=0,title="Class Distribution(0 = Absence , 1 = Presence)"
)
plt.xlabel("HEART DISEASE")
plt.ylabel("Number of Persons")
plt.show()
```

```python
#HEART DISSEASE VS CHOLESTEROL
sns.boxplot(data=df, x='Heart Disease', y='Cholesterol')

plt.title("Cholesterol vs Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Cholesterol")
plt.show()
```

```python
#Separate Features and Target
X = df[["Age","Sex", "Chest pain type","BP" ,"Cholesterol","FBS over 120" ,"EKG results", "Max HR","Exercise angina","ST depression","Slope of ST","Number of vessels fluro"]]
Y = df["Heart Disease"]

print("Features(X): ")
print(X.head())
print("Target(y): ")
print(Y.head())
```

```python
#Train-Test Split

X_train,X_test,Y_train,Y_test=train_test_split(
    X,Y,
    test_size= 0.25,
    random_state=42,
    stratify=Y
)
```

```python
#Feature scaling and Model Creation

model= Pipeline([
    ("scaler",StandardScaler()),
    ("classifier",LogisticRegression())
])
```

```python
#Train the Model
model.fit(X_train,Y_train)
print("Model Training Completed ")
```

```python
#Predict the test Data
Y_pred=model.predict(X_test)
comparison=pd.DataFrame({
    "Actual_Result":Y_test.values,
    "Predicted_Result":Y_pred
})
print(comparison)
```

```python
#Confusion Matrix
cm=confusion_matrix(Y_test,Y_pred,labels=['Absence', 'Presence'])
print("Rows= Actual, Columns= Predicted (Absence, Presence)")
print(cm)
```

```python
#Calculate Evaluation Matrics

accuracy=accuracy_score(Y_test,Y_pred)
precision=precision_score(Y_test,Y_pred,pos_label='Presence',zero_division=0)
recall=recall_score(Y_test,Y_pred,pos_label='Presence',zero_division=0)
f1=f1_score(Y_test,Y_pred,pos_label='Presence',zero_division=0)

print("Accuracy: ",round(accuracy,3))
print("Precision (for Presence): ",round(precision,3))
print("Recall (for Presence): ",round(recall,3))
print("F1 Score (for Presence): ",round(f1,3))

print("\n Detailed Classification Report: ")
print(classification_report(Y_test,Y_pred,
                            labels=['Absence','Presence'],
                            target_names=["Absence","Presence"],
                            zero_division=0))
```

```python
# Understand Prediction Probability


probabilities = model.predict_proba(X_test)
print("Columns are ordered as classes : ",model.classes_)
print(probabilities)
```

```python
#Predict for a New Patient
new_patient=pd.DataFrame({
    "Age":[60],
    "Sex":[1],
    "Chest pain type":[3],
    "BP":[140],
    "Cholesterol":[289],
    "FBS over 120":[1],
    "EKG results":[2],
    "Max HR":[172],
    "Exercise angina":[0],
    "ST depression":[1],
    "Slope of ST":[2],
    "Number of vessels fluro":[2]
})
predicted_class=model.predict(new_patient)[0]
Predicted_Probability=model.predict_proba(new_patient)[0]
if predicted_class == 'Presence':
  print ("Predicted Result: Presence")
else:
  print("Predicted Result: Absence")
print("Class Order:",model.classes_)
print("Predicted Probabilities:",Predicted_Probability)
```
